```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Dataset

data = [
    ["Name", "Age", "City"],
    ["Alice", 30, "New York"],
    ["Bob", 25, "Los Angeles"],
    ["Carol", 27, "Chicago"],
    ["David", 22, "Miami"],
    ["Eva", 29, "Seattle"],
    ["Frank", 31, "Boston"],
    ["Grace", 24, "San Francisco"],
    ["Hannah", 26, "Austin"],
    ["Ian", 28, "Denver"],
    ["Jack", 23, "Phoenix"],
    ["Kathy", 32, "Atlanta"],
    ["Leo", 29, "Dallas"],
    ["Mia", 25, "Orlando"],
    ["Nina", 27, "Philadelphia"],
    ["Oscar", 30, "San Diego"],
    ["Paula", 22, "Las Vegas"],
    ["Quinn", 28, "Portland"],
    ["Rita", 26, "Houston"],
    ["Sam", 24, "Charlotte"],
    ["Tina", 29, "Indianapolis"],
    ["Uma", 31, "Nashville"]
]
```

```python
# Headers and Rows
headers = data[0]
rows = data[1:]
```

```python
df = pd.DataFrame(rows, columns=headers)
```

```python
df.describe()
```

```python
print(df)
```

```python
highest_age = df['Age'].max()
print(highest_age)
```

```python
lowest_age = df['Age'].min()
print(lowest_age)
```

```python
# Create Box plot
plt.boxplot(df['Age'])
plt.title('Box Plot of Age')
plt.ylabel('Age')
plt.show()
```

```python
# Create box plot
plt.figure(figsize=(20, 10))
sns.boxplot(x='Name', y='Age', data=df)
plt.title('Box Plot of Age by Name')
plt.ylabel('Age')
plt.show()
```

```python
# Create box plot
plt.figure(figsize=(15, 5))
sns.barplot(x='Name', y='Age', data=df)
plt.title('Box Plot of Age by Name')
plt.ylabel('Age')
plt.show()
```

```python
plt.figure(figsize=(15, 5))
sns.lineplot(x='Name', y='Age', data=df)
plt.title('Line Plot of Age by Name')
plt.ylabel('Age')
plt.show()
```

```python
plt.figure(figsize=(15, 5))
sns.scatterplot(x='Name', y='Age', data=df)
plt.title('Scatter Plot of Age by Name')
plt.ylabel('Age')
plt.show()
```
