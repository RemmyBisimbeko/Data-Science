```python
import pandas as pd
import numpy as np
```

```python
# Install all necessary libraries
import seaborn as sb 
import matplotlib.pyplot as plt

# load the dataset
data = pd.read_excel('/Users/remmy/Documents/Projects/Data-Science/SEM 1/Data Sets/Uganda Coffee_Cupping.xlsx')
data.head()
```

```python
# Check for duplicates
data.duplicated().sum()

data.drop_duplicates(inplace=True) 
data.duplicated().sum()
```

```python
# Add this line to a new cell to see all column names
print(data.columns)
import seaborn as sb
import matplotlib.pyplot as plt

# Corrected code using the actual column name 'ORIGIN'
plt.figure(figsize=(10,6))
sb.histplot(data['ORIGIN'], bins=30, kde=True, color='brown')
plt.title('Distribution of Coffee Origin')
plt.xlabel('Origin')
plt.ylabel('Count')
plt.show()
# --- Data Cleaning Step ---

# Print original column names
print("Original columns:\n", data.columns.values)

# Clean the column names
data.columns = (data.columns
                .str.strip()                 # Remove leading/trailing whitespace
                .str.lower()                 # Convert to lowercase
                .str.replace(' ', '_')       # Replace spaces with underscores
                .str.replace('/', '_')       # Replace slashes with underscores
                .str.replace('[^a-z0-9_]', '', regex=True) # Remove any other non-alphanumeric characters
               )

# Print the new, cleaned column names
print("\nCleaned columns:\n", data.columns.values)

# --- Now you can plot using the clean name ---
plt.figure(figsize=(10,6))
sb.histplot(data['origin'], bins=30, kde=True, color='brown') # Use the new, clean name
plt.title('Distribution of Coffee Origin')
plt.xlabel('Origin')
plt.ylabel('Count')
plt.show()
```
